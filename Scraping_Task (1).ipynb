```python
import selenium 
import pandas as pd
from selenium import webdriver
import warnings
warnings.filterwarnings('ignore')
from selenium.webdriver.common.by import By
import time
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import requests
```

```python
driver = webdriver.Chrome()

#Opening Car's24.com Page an autometed chrome browser.

driver.get("https://www.cars24.com/")
```

```python
search = driver.find_element(By.XPATH,"/html/body/div[1]/div[2]/div/div/div[1]/div[1]/div")

search.click()
```

```python
# scrape all product urls

product_urls = []

start = 0
end = 3

for page in range(start, end):  # for Loop for scrapping 3 pages
    urls = driver.find_elements(By.XPATH, '//div[@class="_2ujGx"]/a')

    for i in urls:
        # scraping urls
        product_urls.append(i.get_attribute("href"))

    # clicking the next button
    #nxt_button = driver.find_element(By.XPATH, '//span[@class="s-pagination-strip"]/a')
    #nxt_button.click()

    time.sleep(2)
```

```python
cars_urls = product_urls
```

```python
car_make_list = []
model_list = []
km_driven_list = []
fuel_type_list = []
transmission_type_list = []
price_list = []
location_list = []
```

```python

for url in cars_urls:  # loop for every car in the list
    driver.get(url)
    time.sleep(2)
    
    try:
        car_make = driver.find_element(By.XPATH, '//span[@class="car-make-class"]')  # Replace with the correct XPath
        car_make_list.append(car_make.text)
    except NoSuchElementException:
        car_make_list.append('-')
        
    try:
        model = driver.find_element(By.XPATH, '//span[@class="model-class"]')  # Replace with the correct XPath
        model_list.append(model.text)
    except NoSuchElementException:
        model_list.append('-')
        
    try:
        km_driven = driver.find_element(By.XPATH, '//span[@class="km-driven-class"]')  # Replace with the correct XPath
        km_driven_list.append(km_driven.text)
    except NoSuchElementException:
        km_driven_list.append('-')
    
    try:
        fuel_type = driver.find_element(By.XPATH, '//span[@class="fuel-type-class"]')  # Replace with the correct XPath
        fuel_type_list.append(fuel_type.text)
    except NoSuchElementException:
        fuel_type_list.append('-')
        
    try:
        transmission_type = driver.find_element(By.XPATH, '//span[@class="transmission-type-class"]')  # Replace with the correct XPath
        transmission_type_list.append(transmission_type.text)
    except NoSuchElementException:
        transmission_type_list.append('-')
        
    try:
        price = driver.find_element(By.XPATH, '//span[@class="price-class"]')  # Replace with the correct XPath
        price_list.append(price.text)
    except NoSuchElementException:
        price_list.append('-')    
        
    try:
        location = driver.find_element(By.XPATH, '//span[@class="location-class"]')  # Replace with the correct XPath
        location_list.append(location.text)
    except NoSuchElementException:
        location_list.append('-')  
```

```python
car_data = ['2017 Maruti Swift VXI', '2016 Maruti Alto K10 VXI AMT', '2017 Maruti IGNIS DELTA 1.2 AMT', 
            '2014 Maruti Wagon R 1.0 LXI CNG', '2018 Maruti Celerio VXI CNG', '2020 Maruti Celerio VXI (O) CNG', 
            '2022 Maruti Baleno ALPHA 1.2 AGS', '2022 Maruti Baleno DELTA PETROL 1.2', '2013 Maruti Wagon R 1.0 VXI', 
            '2017 Maruti Wagon R 1.0 LXI CNG', '2021 Maruti Celerio VXI', '2019 Maruti Celerio VXI CNG', 
            '2018 Maruti IGNIS SIGMA 1.2', '2021 Maruti Baleno DELTA CVT PETROL 1.2', '2016 Maruti Swift VXI', 
            '2023 Maruti BREZZA VXI SMART HYBRID', '2011 Maruti Swift Dzire ZXI']

# Initialize lists to hold separated data
car_names = []
car_models = []

for car in car_data:
    parts = car.split()
    year = parts.pop(0)  # Remove the year
    car_name = ' '.join(parts[:2])  # Join the first two parts as car name
    car_model = ' '.join(parts[2:])  # Join the remaining parts as the car model
    
    car_names.append(car_name)
    car_models.append(car_model)

# Output the segregated data
for name, car_model in zip(car_names, car_models):
    print(f"Car Name: {name}, Car Model: {car_model}")

# If you want to store the segregated data in a list of dictionaries
segregated_data = [{"Car Name": name, "Car Model": car_model} for name, car_model in zip(car_names, car_models)]

print(segregated_data)
```
